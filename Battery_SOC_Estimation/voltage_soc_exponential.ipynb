```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn import linear_model
```

```python
df_exp = pd.read_csv('voltage_soc_exponential.csv')
plt.figure(figsize=(10, 5))
plt.title('SOC vs Voltage') 
plt.xlabel('voltage_v')
plt.ylabel('soc_percent')
plt.scatter(df_exp['voltage_v'], df_exp['soc_percent'], label='SOC vs Voltage Exponential', color='red', alpha=0.5,marker='o')
plt.legend()
plt.show()
```

```python
#test data to plot
vtg_s = [2.4, 2.8, 3.0, 3.3, 3.4, 3.5, 3.6, 3.7, 3.8, 3.9, 4.1,4.3]
#exponential
reg_exp = linear_model.LinearRegression()
exp = np.exp(df_exp[['voltage_v']])
reg_exp.fit(exp, df_exp['soc_percent'])
df_predict = pd.DataFrame(vtg_s, columns=['voltage_v'])
soc_s_exp = reg_exp.predict(np.exp(df_predict))
```

```python
plt.figure(figsize=(10, 5))
plt.title('SOC vs Voltage') 
plt.xlabel('voltage_v')
plt.ylabel('soc_percent')

plt.plot(vtg_s, soc_s_exp,color='red',   label='Predict SOC vs Vtg Exponential')

plt.scatter(df_exp['voltage_v'], df_exp['soc_percent'], label='SOC vs Vtg Exponential', color='red' , alpha=0.5,)

plt.legend()
plt.show()
```
